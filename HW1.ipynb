## Solution 1: Sigmoid

Using the chain rule,

$$
\begin{aligned}
\sigma'(x)
&= \frac{d}{dx}(1+e^{-x})^{-1} \\
&= -(1+e^{-x})^{-2}(-e^{-x}) \\
&= \frac{e^{-x}}{(1+e^{-x})^2}.
\end{aligned}
$$

Since

$$
1-\sigma(x)=1-\frac{1}{1+e^{-x}}=\frac{e^{-x}}{1+e^{-x}},
$$

we obtain

$$
\sigma(x)(1-\sigma(x))
=\frac{1}{1+e^{-x}}\cdot\frac{e^{-x}}{1+e^{-x}}
=\frac{e^{-x}}{(1+e^{-x})^2}
=\sigma'(x).
$$

Therefore,

$$
\boxed{\sigma'(x)=\sigma(x)(1-\sigma(x))}.
$$

## Solution 2: Tanh

The derivatives of the numerator and denominator are

$$
\frac{d}{dx}(e^x-e^{-x})=e^x+e^{-x},
\qquad
\frac{d}{dx}(e^x+e^{-x})=e^x-e^{-x}.
$$

Using the quotient rule,

$$
\begin{aligned}
\frac{d}{dx}\tanh(x)
&=\frac{(e^x+e^{-x})^2-(e^x-e^{-x})^2}{(e^x+e^{-x})^2} \\
&=1-\left(\frac{e^x-e^{-x}}{e^x+e^{-x}}\right)^2 \\
&=1-\tanh^2(x).
\end{aligned}
$$

Therefore,

$$
\boxed{\frac{d}{dx}\tanh(x)=1-\tanh^2(x)}.
$$
